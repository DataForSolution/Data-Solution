# Spark/Glue ingestion-validation demo

This public notebook uses a tiny Spark-like test double to demonstrate the historical embedded-header failure. It does not connect to AWS or include credentials.

```python
from spark_portfolio.validation import dataframe_audit

class Row(dict):
    def asDict(self, recursive=True): return dict(self)

class Frame:
    columns = ['sk_id_curr', 'target', 'name_contract_type', 'code_gender']
    rows = [Row(sk_id_curr=None, target=None, name_contract_type='NAME_CONTRACT_TYPE', code_gender='CODE_GENDER')]
    def count(self): return len(self.rows)
    def first(self): return self.rows[0]

dataframe_audit(Frame())
```
