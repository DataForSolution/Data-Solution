# PETQuant safety-metric demo

This public notebook uses **synthetic labels and scores only**. It demonstrates the false-reliable metric and a conservative confidence gate; it does not estimate clinical performance.

```python
from petquant_portfolio.safety import false_reliable_rate, apply_reliable_score_gate
```

## Synthetic example

The values below are deliberately small and interpretable so the safety behavior can be inspected directly.

```python
reference = ["unreliable", "unreliable", "caution", "reliable", "reliable"]
predicted = ["reliable", "caution", "reliable", "reliable", "reliable"]
scores = [82.0, 60.0, 75.0, 94.0, 88.0]
false_reliable_rate(reference, predicted)
```

```python
gated = apply_reliable_score_gate(predicted, scores, min_reliable_score=90.0)
gated
```

```python
false_reliable_rate(reference, gated)
```

## Interpretation

A gate can reduce false reassurance by converting low-confidence `reliable` predictions to `caution`. That safety trade-off must be evaluated on independent, appropriately labeled data before any threshold is treated as validated.
