# Pima diabetes evaluation-contract demo

This notebook uses a tiny synthetic Pima-shaped frame to demonstrate leakage-safe preprocessing and metric reporting. It does not estimate clinical performance.

```python
import pandas as pd
from pima_diabetes_portfolio.data import prepare_pima_frame
from pima_diabetes_portfolio.modeling import build_svm_pipeline
from pima_diabetes_portfolio.evaluation import classification_report_dict

frame = pd.DataFrame({
    'preg':[0,1,2,3,1,4,2,5], 'plas':[90,140,0,130,85,160,100,150],
    'pres':[70,80,0,76,68,82,72,78], 'skin':[20,30,0,25,18,35,22,32],
    'insu':[80,120,0,100,70,180,90,150], 'mass':[25,32,0,29,24,36,27,34],
    'pedi':[.2,.6,.3,.5,.1,.8,.4,.7], 'age':[22,45,33,40,24,51,30,48],
    'class':['tested_negative','tested_positive']*4,
})
X, y = prepare_pima_frame(frame)
model = build_svm_pipeline(kernel='linear')
model.fit(X, y)
pred = model.predict(X)
scores = model.predict_proba(X)[:,1]
classification_report_dict(y, pred, positive_scores=scores)
```

## Interpretation boundary

The synthetic example verifies mechanics only. A real experiment should select models on train/validation data and evaluate once on held-out OpenML-37 test data.
