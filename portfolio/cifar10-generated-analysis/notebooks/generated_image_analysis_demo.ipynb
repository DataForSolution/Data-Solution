# Generated-image classifier-response analysis

This notebook uses a small deterministic probability matrix to demonstrate the corrected analysis contract. Classifier confidence is **not** treated as image realism.

```python
import numpy as np
from cifar10_portfolio.analysis import prediction_table, class_confidence_summary, class_coverage_summary, top_confidence_indices

class_names = ['airplane', 'automobile', 'bird', 'cat']
probabilities = np.array([
    [0.70, 0.10, 0.10, 0.10],
    [0.05, 0.80, 0.10, 0.05],
    [0.10, 0.10, 0.65, 0.15],
    [0.15, 0.10, 0.20, 0.55],
    [0.50, 0.20, 0.20, 0.10],
])
prediction_table(probabilities, class_names)
```

```python
class_confidence_summary(probabilities, class_names)
```

```python
class_coverage_summary(probabilities, class_names), top_confidence_indices(probabilities, n=3)
```

## Interpretation

High confidence and broad predicted-class coverage can be useful diagnostics of a generator/classifier pair, but neither establishes perceptual realism or generative fidelity.
